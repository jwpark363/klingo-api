```python
import pandas as pd
import requests,json
from db.model.interview import InterviewCreate
from pydantic import BaseModel
BASE_URL = "http://localhost:8104/"
```

```python
def insert_interview(id, item):
    ## 인터뷰 데이터 입력
    api = f"{BASE_URL}/interview/post"
    response = requests.post(api, data=item.model_dump_json())
    print(response.status_code)
    if response.status_code != 201:
        print(id,response.json())
def sample_interview(token):
    ## 인터뷰 데이터 가져오기(랜덤하게 5게)
    api = f"{BASE_URL}/interview/hello"
    header = {
        "Authorization":f"Bearer {token}"
    }
    print(header)
    response = requests.get(api, headers=header)
    print(response.status_code)
    if response.status_code != 201:
        return response.json()
    return None

class Answer(BaseModel):
    interview_id: int
    answer: str
    
def insert_interview_answer(token,answers:list[Answer]):
    ## 인터뷰 답변 데이터 입력(1개 이상)
    api = f"{BASE_URL}/interview/answer/post"
    header = {
        "Authorization":f"Bearer {token}"
    }
    _answers = []
    for answer in answers:
        _answers.append({
            'interview_id':answer.interview_id,
            'answer':answer.answer,
            'user_id':None
        })
    print(_answers)
    response = requests.post(api, data=json.dumps(_answers), headers=header)
    print(response.status_code)
    if response.status_code != 201:
        print(id,response.json())
    return response.json()
```

```python
## 샘플 answer 입력 테스트
## 1. 토큰 생성
api = f"{BASE_URL}/users/token"
body = {
    "username":"klingo_admin",
    "password":"administration"
}
response = requests.post(api, data=body)
token = response.json()["access_token"]
print(token)
datas = [
    Answer(interview_id=1,answer='test'),
    Answer(interview_id=2,answer='test')
]
result = insert_interview_answer(token,datas)
```
